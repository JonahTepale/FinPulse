```python
# ============================================================
# FinPulse — Data Cleaning Script
# Dataset: Credit Card Transactions Dataset (Kaggle)
# Run each section in order inside Jupyter Notebook
# ============================================================
 
import pandas as pd
import numpy as np
 
# ============================================================
# CELL 1 — Load the data
# ============================================================
 
df = pd.read_csv('data/credit_card_transactions.csv')  # update filename if different
 
print(f"Shape: {df.shape}")
print(f"\nColumns:\n{df.columns.tolist()}")
print(f"\nFirst 5 rows:")
df.head()
 
```

```python
# ============================================================
# CELL 2 — Snapshot: understand what you're working with
# ============================================================
 
print("=== DATA TYPES ===")
print(df.dtypes)
 
print("\n=== MISSING VALUES ===")
print(df.isnull().sum())
 
print("\n=== BASIC STATS ===")
print(df.describe())
 
```

```python
# ============================================================
# CELL 3 — Drop duplicates
# ============================================================
 
before = len(df)
df = df.drop_duplicates()
after = len(df)
 
print(f"Removed {before - after} duplicate rows. {after} rows remaining.")
```

```python
# ============================================================
# CELL 4 — Handle missing values
# ============================================================
 
# Drop rows where amount or date is missing (non-negotiable fields)
df = df.dropna(subset=['amt', 'trans_date_trans_time'])  # adjust column names if needed
 
# Fill missing merchant names with 'Unknown'
if 'merchant' in df.columns:
    df['merchant'] = df['merchant'].fillna('Unknown')
 
# Fill missing category with 'other'
if 'category' in df.columns:
    df['category'] = df['category'].fillna('other')
 
print(f"Missing values after cleaning:\n{df.isnull().sum()}")
```

```python
# ============================================================
# CELL 5 — Fix data types
# ============================================================
 
# Parse transaction date into datetime
df['trans_date'] = pd.to_datetime(df['trans_date_trans_time'], errors='coerce')
 
# Extract useful time features
df['year']  = df['trans_date'].dt.year
df['month'] = df['trans_date'].dt.month
df['day_of_week'] = df['trans_date'].dt.day_name()
 
# Make sure amount is numeric
df['amt'] = pd.to_numeric(df['amt'], errors='coerce')
 
print("Date range:", df['trans_date'].min(), "→", df['trans_date'].max())
print("Sample amounts:", df['amt'].describe())
```

```python
# ============================================================
# CELL 6 — Remove outliers (extreme transaction amounts)
# ============================================================
 
# Use IQR method to flag outliers
Q1 = df['amt'].quantile(0.25)
Q3 = df['amt'].quantile(0.75)
IQR = Q3 - Q1
 
lower = Q1 - 3 * IQR
upper = Q3 + 3 * IQR
 
outliers = df[(df['amt'] < lower) | (df['amt'] > upper)]
print(f"Outliers found: {len(outliers)} rows")
print(outliers[['trans_date', 'amt', 'category']].head(10))
 
# Remove them
df = df[(df['amt'] >= lower) & (df['amt'] <= upper)]
print(f"Rows after outlier removal: {len(df)}")
```

```python
# ============================================================
# CELL 7 — Standardize category labels
# ============================================================
 
print("Raw categories:")
print(df['category'].value_counts())
 
# Normalize: lowercase + strip whitespace
df['category'] = df['category'].str.lower().str.strip()
 
# Optional: remap categories to cleaner labels
category_map = {
    'food_dining'         : 'dining',
    'grocery_pos'         : 'groceries',
    'grocery_net'         : 'groceries',
    'shopping_net'        : 'shopping',
    'shopping_pos'        : 'shopping',
    'gas_transport'       : 'transport',
    'travel'              : 'travel',
    'entertainment'       : 'entertainment',
    'health_fitness'      : 'health',
    'personal_care'       : 'personal care',
    'home'                : 'home',
    'kids_pets'           : 'kids & pets',
    'misc_net'            : 'miscellaneous',
    'misc_pos'            : 'miscellaneous',
}
 
df['category_clean'] = df['category'].map(category_map).fillna(df['category'])
 
print("\nCleaned categories:")
print(df['category_clean'].value_counts())
```

```python
# ============================================================
# CELL 8 — Build monthly summary table (for scoring model)
# ============================================================
 
monthly = df.groupby(['year', 'month']).agg(
    total_spent    = ('amt', 'sum'),
    avg_transaction= ('amt', 'mean'),
    num_transactions=('amt', 'count'),
).reset_index()
 
monthly['month_label'] = pd.to_datetime(
    monthly[['year', 'month']].assign(day=1)
).dt.strftime('%b %Y')
 
print(monthly.tail(10))
 
```

```python
# ============================================================
# CELL 9 — Build category breakdown table
# ============================================================
 
category_summary = df.groupby('category_clean').agg(
    total_spent     = ('amt', 'sum'),
    avg_transaction = ('amt', 'mean'),
    num_transactions= ('amt', 'count'),
).reset_index().sort_values('total_spent', ascending=False)
 
print(category_summary)
```

```python
# ============================================================
# CELL 10 — Save cleaned files
# ============================================================
 
df.to_csv('finpulse_transactions_clean.csv', index=False)
monthly.to_csv('finpulse_monthly_summary.csv', index=False)
category_summary.to_csv('finpulse_category_summary.csv', index=False)
 
print("\n✅ Cleaning complete. 3 files saved:")
print("  → finpulse_transactions_clean.csv   (main cleaned dataset)")
print("  → finpulse_monthly_summary.csv      (monthly KPIs)")
print("  → finpulse_category_summary.csv     (spending by category)")
```
